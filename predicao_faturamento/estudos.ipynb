# Estudos predição faturamento

```python
import pandas as pd
from pathlib import Path
import numpy as np
import matplotlib.pyplot as pt

dataset = Path.joinpath(Path('__file__').parent, 'dataset', 'df.csv')
```

## Carregamento dos dados

```python

df = pd.read_csv(dataset).round(2)

df.head()
```

## Análise do dataset

```python
df['Mes'].unique()
```

## Média aritmética

```python
media = df['Faturamento'].mean()
media.round(2)
```

## Desvio Padrão

```python
devpad = df['Faturamento'].std()
devpad.round(2)
```

## Verificando a previsão de faturamento 

### Comparativo em tabela

```python
colunas = ['Min', 'Média', 'Máx']
min = media - devpad
max = media + devpad

valores = np.array([min, media, max])
valores
```

```python
df_media = pd.DataFrame([valores], columns=colunas).round(2)
df_media
```

### Coeficiente de variação do faturamento

```python
coe_var = ((devpad/media)*100).round(2)

coe_var_str = 'Faturamento pode variar em torno de '+str(coe_var)+'%'
coe_var_str
```

### Verificação da variação com gráfico

```python
pt.plot(df['Faturamento'], marker='o')
```

## Média Ponderada (meses com mais peso que outros)

```python
lista = np.transpose((np.array([df['Faturamento'].tail(), np.arange(1,6)])))
pesos = np.arange(1,6)

df_ult = pd.DataFrame(lista, columns=['Ultimos','Pesos'])

df_ult['Ponderado'] = df_ult['Ultimos']*df_ult['Pesos']

df_ult
```

```python
med_pond = df_ult['Ponderado'].sum()/df_ult['Pesos'].sum()

med_pond
```

## Previsão de faturamento 

```python
df_pond = pd.DataFrame(df['Faturamento'].tail(),columns=['Faturamento'])

df_pond.index = np.arange(1, len(df_pond)+1) #reset do index para o valor 1

contador = 0

while contador <=2:
    i = 1
    peso = 0
    soma = 0

    for fat in df_pond['Faturamento'].tail():
        soma += (fat*i)
        peso += i
        i += 1
    
    predicao = soma/peso

    df_pond.loc[df_pond.shape[0]+1] = [predicao]

    contador += 1

df_pond.tail().round(2)
```
